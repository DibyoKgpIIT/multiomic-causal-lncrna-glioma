```python
RNA_symbol_dict = {}
for ind,symbol in enumerate(RNA_symbol_set):
    RNA_symbol_dict[symbol] = ind
RNA_symbol_dict
```

```python
interactor1_index = []
interactor2_index = []
for ind in range(len(df_RNA_combined_short.index)):
    interactor1_index.append(RNA_symbol_dict[df_RNA_combined_short["Interactor1"].loc[ind]])
    interactor2_index.append(RNA_symbol_dict[df_RNA_combined_short["Interactor2"].loc[ind]])
    if(ind%10000==0):
        print(ind,interactor1_index[ind],interactor2_index[ind])
df_RNA_combined_indices = pd.DataFrame(data=list(zip(interactor1_index,interactor2_index)),
                                      columns=["Interactor1.Index","Interactor2.Index"])
df_RNA_combined_indices
```
